```python
# eda_visualization.ipynb
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load data
df = pd.read_csv("https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DV0101EN-SkillsNetwork/Data%20Files/historical_automobile_sales.csv")
df['Date'] = pd.to_datetime(df['Date'])
df['Month'] = df['Date'].dt.month
df['Year'] = df['Date'].dt.year

# 1. Line plot - Yearly sales trend
plt.figure(figsize=(10, 5))
df.groupby("Year")["Automobile_Sales"].sum().plot()
plt.title("Total Automobile Sales by Year")
plt.xlabel("Year")
plt.ylabel("Sales")
plt.grid(True)
plt.show()

# 2. Pie chart - Ad spend by recession
ad_spend = df.groupby("Recession")["Advertising_Expenditure"].sum()
ad_spend.plot.pie(labels=["Non-Recession", "Recession"], autopct="%1.1f%%", figsize=(6, 6), title="Ad Expenditure by Recession")
plt.ylabel("")
plt.show()
```
