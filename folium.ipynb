```python
# folium_map.ipynb
import folium

# Example coordinates (New York)
map_center = [40.7128, -74.0060]

# Create map
m = folium.Map(location=map_center, zoom_start=5)

# Add marker
folium.Marker(location=map_center, popup="Example Dealer", tooltip="NYC Region").add_to(m)

# Save map to file
m.save("automobile_sales_map.html")
m
```
