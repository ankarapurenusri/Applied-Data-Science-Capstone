```python
# sql_analysis.ipynb
import pandas as pd

# Load dataset
df = pd.read_csv("https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DV0101EN-SkillsNetwork/Data%20Files/historical_automobile_sales.csv")

# Extract year and month from date
df['Date'] = pd.to_datetime(df['Date'])
df['Month'] = df['Date'].dt.month
df['Year'] = df['Date'].dt.year

# 1. Average sales by vehicle type
avg_sales = df.groupby("Vehicle_Type")["Automobile_Sales"].mean().reset_index()
print("Average Sales by Vehicle Type:\n", avg_sales)

# 2. Average GDP by year
gdp_trend = df.groupby("Year")["GDP"].mean().reset_index()
print("Average GDP by Year:\n", gdp_trend)
```
